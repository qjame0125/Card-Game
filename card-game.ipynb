```python
import random


class Card:
    def __init__(self, suit: str, rank: str):
        self.suit = suit
        self.rank = rank
    def __repr__(self):
        return f"{self.rank} of {self.suit}"
class Deck:
    def __init__(self):
        self.cards = []
        self.create_deck()
    def create_deck(self):
        suits = ['Heart', 'Diamond', 'Spades', 'Club']
        ranks = ['King', 'Queen', 'Jack', 'Ace','2', '3', '4', '5', '6', '7', '8', '9', '10']
        self.cards = [Card(suit, rank) for suit in suits for rank in ranks]
        random.shuffle(self.cards)
        
    def draw(self):
        return self.cards.pop()
class Player:
    def __init__(self, name):
        self.name = name
        self.hand = []
    def add_card(self, deck):
        card = deck.draw()
        self.hand.append(card)
        return card
    
class Gamerule:
    @staticmethod
    def card_values(card):
        face_card = ['King', 'Queen', 'Jack',]
        if card.rank in face_card:
            return 10
        elif card.rank == "Ace":
            return 11  
        return int(card.rank)
    @staticmethod
    def total_score(hand):
        score = 0
        aces = 0
        for card in hand:
            if card.rank == "Ace":
                aces += 1
                score += 11
            else:
                score += Gamerule.card_values(card)
        while score > 21 and aces > 0:
            score -= 10
            aces -= 1
        return score
    def player_turn(player, deck):
        while True:
            score = Gamerule.total_score(player.hand)
            print(f"\n{player.name}'s hand: {player.hand} Score {score}")
            if score == 21 and len(player.hand) == 2:
                print(f"You Win!")
                return "blackjack"
            if score > 21:
                print(f"You Lose!")
                return "bust"
            action = input("Hit (h) or Stand(s)?").lower()
            if action == "h":
                player.add_card((deck))
            elif action == "s":
                print(f"{player.name} stands")
                return "stand"
            else:
                print("Invalid choice. Please choose Hit(h) or Stand(s)")
    @staticmethod
    def cpu_turn(cpu, deck):
        print(f"\n {cpu.name}'s turn")
        while True:
            score = Gamerule.total_score(cpu.hand)
            print(f"{cpu.name}'s hand: {cpu.hand} Score: {score}")
            if score >= 17:
                print(f"{cpu.name} stands.")
                return score
            if score > 21:
                print(f"{cpu.name} busts!")
                return score
            cpu.add_card(deck)
            print(f"{cpu.name} hits")
if __name__ == "__main__":
    deck = Deck()
    player = Player("Player")
    cpu = Player("Dealer")
    for _ in range(2):
        player.add_card(deck)
        cpu.add_card(deck)
    player_result = Gamerule.player_turn(player, deck)
    player_score = Gamerule.total_score(player.hand)
    if player_score <= 21:
        cpu_score = Gamerule.cpu_turn(cpu, deck)
    else:
        cpu_score = 0
print("\nFinal Score:")
print(f"Player: {player_score} Hand: {player.hand}")
print(f"Dealer:{cpu_score} Hand: {cpu.hand}")
if player_score > 21:
    print("Player busts! Dealer wins.")
elif cpu_score > 21 or player_score > cpu_score:
    print("Player wins!")
else:
    print("It's a tie")
    
    #d = Deck
#d.create_deck()
```
